```python
print("=== PROGRAM KONVERSI SUHU CELCIUS KE FAHRENHEIT ===")

# 1. Ambil input suhu Celsius
celcius = float(input("Masukkan suhu dalam derajat Celcius: "))

# 2. Hitung konversi ke Fahrenheit
fahrenheit = (celcius*1.8) + 32

# 3. Tampilkan hasilnya
print(f"Suhu {celcius}°C sama dengan {fahrenheit}°F")
```
